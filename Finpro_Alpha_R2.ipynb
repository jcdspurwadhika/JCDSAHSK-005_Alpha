**Metric Evaluation**

| Actual / Predicted | Not Canceled (0) | Canceled (1) |
|---|---|---|
| **Not Canceled (0)** | **True Negative (TN)** | **False Positive (FP)** |
| **Canceled (1)** | **False Negative (FN)** | **True Positive (TP)** |

**Confusion Matrix Interpretation**

*True Positive (TP)*

A booking is canceled and the model correctly predicts cancellation.

*True Negative (TN)*

A booking is not canceled and the model correctly predicts no cancellation.

*False Positive (FP)*

A booking is not canceled but the model predicts cancellation.

Business impact:
Additional effort may be spent contacting customers who would not actually cancel.

*False Negative (FN)*

A booking is canceled but the model predicts no cancellation.

Business impact:
The hotel loses the opportunity to intervene and may lose potential revenue.

**Business Cost Perspective**

In hotel operations, not all prediction errors generate the same business impact.

***False Positive (FP)***

A booking is predicted to cancel but the customer actually completes the reservation.

Business Impact

*   Additional follow-up effort
*   Unnecessary intervention cost
*   Additional operational workload


***False Negative (FN)***

A booking is predicted not to cancel but the customer actually cancels.

Business Impact

*   Lost revenue opportunity
*   Missed intervention opportunity
*   Reduced occupancy utilization
*   Inaccurate revenue forecasting

Because False Negatives are directly associated with potential revenue loss, they are considered more expensive than False Positives.

Therefore, the business cost framework can be represented as:

    False Positive Cost = Number of False Positives × Follow-Up Cost

    False Negative Cost = Number of False Negatives × Revenue Loss per Cancellation

    Total Business Cost = False Positive Cost + False Negative Cost

At this stage, the formulas are introduced conceptually to explain the business impact of prediction errors. Actual business cost calculations will be performed later after model evaluation and confusion matrix results become available.

## Business Cost Evaluation

In a hotel booking cancellation prediction system, different prediction errors generate different levels of business impact.

A **False Positive (FP)** occurs when the model predicts that a booking will be cancelled, but the guest actually checks in. Although this leads to unnecessary operational actions, such as reminder emails, reservation reconfirmation, or follow-up calls, the financial impact is relatively limited.

A **False Negative (FN)** occurs when the model predicts that a booking will not be cancelled, but the customer ultimately cancels the reservation. This type of error is substantially more costly because it may result in lost room revenue, reduced occupancy, inaccurate revenue forecasting, and missed mitigation opportunities.

Therefore, a cost-sensitive evaluation framework is applied to quantify the financial impact of model errors.

### Cost Assumptions

For business evaluation purposes, the following assumptions are used:

```text
Follow-Up Cost per False Positive     = $10.00
Revenue Loss per False Negative       = $101.83
```

Where:

- **$10.00** represents the estimated operational cost of conducting follow-up activities.

- **$101.83** represents the Average Daily Rate (ADR) from the hotel booking dataset, used as a proxy for potential revenue loss per missed cancellation.

### Business Cost Formula

```text
False Positive Cost = Number of False Positives × $10.00

False Negative Cost = Number of False Negatives × $101.83

Total Business Cost =
False Positive Cost + False Negative Cost
```

### Example Calculation

Assume the final model at the selected production threshold produces the following results:

```text
False Positives (FP) = 450
False Negatives (FN) = 120
```

Then:

```text
False Positive Cost
= 450 × $10.00
= $4,500.00

False Negative Cost
= 120 × $101.83
= $12,219.60

Total Business Cost
= $4,500.00 + $12,219.60
= $16,719.60
```

### Business Interpretation

The calculation shows that the financial impact of False Negatives is significantly larger than that of False Positives. Therefore, the model prioritizes identifying as many potential cancellations as possible, even if it results in a moderate increase in follow-up activities.

This approach supports the primary business objective of **hotel revenue protection**, where preventing missed cancellation cases is more valuable than avoiding unnecessary customer follow-ups.

Consequently, threshold selection is evaluated not only using traditional machine learning metrics such as Recall, F2-Score, and ROC-AUC, but also through its impact on total business cost.

Primary Metrics:

- Recall
- F2 Score

Secondary Metrics:

- Precision
- Accuracy
- ROC-AUC


    Recall = TP / (TP + FN)

    F2 Score = (1 + 2²) × (Precision × Recall) / ((2² × Precision) + Recall)
    
    Precision = TP / (TP + FP)

    Accuracy = (TP + TN) / (TP + TN + FP + FN)


Why Recall and F2 Score Are Prioritized

The primary objective of this project is to identify bookings that are likely to be cancelled before revenue loss occurs. In this business context, False Negatives are considerably more costly than False Positives because undetected cancellations may result in lost revenue, inaccurate occupancy forecasting, and missed intervention opportunities.

Recall is therefore prioritized because it measures the model's ability to identify actual cancellation cases and directly minimizes False Negatives.

However, Recall alone may encourage excessive intervention activities. To achieve a balance between cancellation detection and operational efficiency, F2 Score is also used. F2 Score assigns greater importance to Recall while still considering Precision, making it more suitable for the business objective of revenue protection.

*Treatment Decision*

Missing values will be replaced with a new category named `"No Agent"`.

Missing Value Treatment Strategy

Based on business understanding and data quality assessment, the following treatments will be applied:

| Feature | Missing Values | Treatment |

````python
# MISSING VALUE TREATMENT
df['is_company_booking'] = df['company'].notnull().astype(int)
df.drop(columns='company', inplace=True)
df['agent'] = df['agent'].fillna('No Agent')
df['agent'] = df['agent'].astype(str)
df['country'] = df['country'].fillna('Unknown')
df['children'] = df['children'].fillna(0)
````

**Treatment of 'Undefined' Categories**

In addition to true missing values (NaN), two categorical features contain an `Undefined` category that represents an unrecorded value:

- `distribution_channel`: [5] records
- `market_segment`: [2] records (both also among the 5 records above)

Although these entries are not stored as NaN, they carry no usable business information, because the booking channel and customer segment are unknown. Since both features are important cancellation indicators in the exploratory analysis, keeping an uninformative category would add noise rather than signal.

These records are **dropped instead of imputed** for the following reasons:

1. **Negligible volume.** Only 5 rows in total (119,390 → 119,385), or about 0.004% of the dataset, so removing them does not materially change the data distribution or the target distribution.
2. **Imputation would be speculative.** Replacing the category with the most frequent value (for example, Online TA) would assign a channel or segment that was never recorded and could introduce bias.
3. **No meaningful impact on modeling.** With such a small share of the data, the effect on model performance is expected to be negligible.

This step is performed before duplicate handling, so all later row counts are based on 119,385 observations.

````python
df = df.drop(
    df[(df['distribution_channel'] == 'Undefined') |
    (df['market_segment'] == 'Undefined')].index
)
````

## **5.2 Feature Engineering**

````python
def create_features(data):

    data = data.copy()

    # Total Nights
    data['total_nights'] = (
        data['stays_in_week_nights']
        + data['stays_in_weekend_nights']
    )

    # Total Guests
    data['total_guests'] = (
        data['adults']
        + data['children']
        + data['babies']
    )

    # Booking Revenue
    data['booking_revenue'] = (
        data['adr']
        * data['total_nights']
    )

    # Net Cancellation History
    data['net_cancelled'] = (
        data['previous_cancellations']
        - data['previous_bookings_not_canceled']
    )

    # Revenue Per Guest
    data['rev_per_guest'] = (
        data['booking_revenue']
        /
        data['total_guests'].replace(0,1)
    )

    return data
````

Note on `arrival_date_year`: it is kept as an input to stay consistent with the benchmark results, but a model that depends on the calendar year cannot extrapolate to future years. This is listed as a limitation in the conclusion.Different preprocessing techniques will be applied to numerical and categorical variables.

````python
# Numerik biasa
standard_cols = [
    'lead_time', 'stays_in_weekend_nights', 'stays_in_week_nights',
    'adults', 'children', 'babies',
    'previous_cancellations', 'previous_bookings_not_canceled',
    'booking_changes', 'days_in_waiting_list',
    'required_car_parking_spaces', 'total_of_special_requests'
]

# Numerik dengan outlier ekstrem
robust_cols = ['adr']

# Kategorikal kardinalitas rendah (assigned_room_type has been removed to avoid leakage/errors)
ohe_cols = [
    'hotel', 'meal', 'market_segment', 'distribution_channel',
    'deposit_type', 'customer_type', 'reserved_room_type',
    'arrival_date_month', 'assigned_room_type'
]

# Kategorikal kardinalitas tinggi
freq_cols = ['country', 'agent']

preprocessor = ColumnTransformer(
    transformers=[
        ('standard', StandardScaler(), standard_cols),
        ('robust', RobustScaler(), robust_cols),
        ('ohe', OneHotEncoder(handle_unknown='ignore'), ohe_cols),
        ('freq', CountEncoder(normalize=True, handle_unknown=0, handle_missing=0), freq_cols)
    ],
    remainder='passthrough'
)

display(preprocessor)
````

## **6.3 Resampling Check: With vs Without Oversampling**

````python
# ==========================================
# RESAMPLING CANDIDATES
# ==========================================

resampling_candidates = {
    'XGBoost': XGBClassifier(
        random_state=42,
        eval_metric='logloss'
    ),
    'LightGBM': LGBMClassifier(
        random_state=42,
        verbose=-1
    )
}

# ==========================================
# RESAMPLING METHODS
# ==========================================

resampler = {
    'No Oversampling': None,
    'Random Over Sampler': RandomOverSampler(random_state=42),
    'SMOTE': SMOTE(random_state=42),
    'SMOTETomek': SMOTETomek(random_state=42)
}

# ==========================================
# EVALUATION METRICS
# Primary Metrics:
# 1. Recall
# 2. F2 Score
#
# Secondary Metrics:
# 3. Precision
# 4. ROC-AUC
# ==========================================

resampling_scoring = {
    'Recall': 'recall',
    'F2': f2_scorer,
    'Precision': 'precision',
    'ROC_AUC': 'roc_auc'
}

# ==========================================
# RESAMPLING BENCHMARK
# ==========================================

resampling_rows = []

for name, mdl in resampling_candidates.items():

    for sampler_name, sampler_obj in resampler.items():

        if sampler_obj is None:

            pipe = ImbPipeline([
                ('preprocessor', preprocessor),
                ('model', mdl)
            ])

        else:

            pipe = ImbPipeline([
                ('preprocessor', preprocessor),
                ('oversampler', sampler_obj),
                ('model', mdl)
            ])

        scores = cross_validate(
            estimator=pipe,
            X=X_train,
            y=y_train,
            cv=cv,
            scoring=resampling_scoring,
            n_jobs=-1
        )

        resampling_rows.append({

            'Model': name,
            'Resampling': sampler_name,

            # PRIMARY METRICS
            'Recall':
            scores['test_Recall'].mean(),

            'F2 Score':
            scores['test_F2'].mean(),

            # SECONDARY METRICS
            'Precision':
            scores['test_Precision'].mean(),

            'ROC AUC':
            scores['test_ROC_AUC'].mean()

        })

# ==========================================
# RESULTS TABLE
# ==========================================

resampling_df = pd.DataFrame(
    resampling_rows
)

resampling_df = (
    resampling_df
    .sort_values(
        by=['F2 Score', 'Recall'],
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    resampling_df[
        [
            'Model',
            'Resampling',
            'Recall',
            'F2 Score',
            'Precision',
            'ROC AUC'
        ]
    ].round(4)
)
````

**Resampling Results Interpretation**

The objective of this project is to identify potential cancellation cases before revenue loss occurs. Therefore, Recall and F2 Score are prioritized when comparing resampling strategies.

Only Random Over Sampler improves cancellation detection substantially: Recall rises by about 0.15 to 0.19 and F2 by about 0.09 to 0.11. SMOTE and SMOTETomek change Recall by only 0.003 to 0.015, and ROC-AUC does not improve with any method (XGBoost 0.9183 to 0.9169, LightGBM 0.9166 to 0.9155).

Random Over Sampling lowers Precision by about 0.11 to 0.13. Part of the Recall gain therefore reflects predictions shifting toward the positive class rather than better class separation. From a business perspective this trade-off is acceptable because missing an actual cancellation (False Negative) is assumed to cost more than an unnecessary follow-up (False Positive).

A side effect is that predicted probabilities become higher than real cancellation frequencies, so they should be read as a ranking signal (see Limitations).

Random Over Sampling is selected as the preferred resampling strategy and is carried forward to hyperparameter tuning.

### **6.4.1 XGBoost**

````python
xgb_params = {
    'model__n_estimators': [200, 300, 400],
    'model__max_depth': [4, 5, 6, 7],
    'model__learning_rate': [0.03, 0.05, 0.08, 0.1],
    'model__subsample': [0.8, 0.9, 1.0],
    'model__colsample_bytree': [0.7, 0.8, 0.9],
    'model__scale_pos_weight': [1],
    'model__min_child_weight': [1, 3, 5],
    'model__reg_lambda': [1, 2, 5]
}

#XGBoost Pipeline
xgb_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('oversampler', RandomOverSampler(random_state=42)),
    ('model',
     XGBClassifier(
         random_state=42,
         eval_metric='logloss'
     ))
])

#Randomized Search
random_search_xgb = RandomizedSearchCV(
    estimator=xgb_pipe,
    param_distributions=xgb_params,
    n_iter=20,
    scoring=f2_scorer,
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=2
)
random_search_xgb.fit(X_train,y_train)
````

### **6.4.4 Soft Voting**

The Soft Voting ensemble combines the tuned XGBoost, Extra Trees and LightGBM models

````python
from sklearn.ensemble import VotingClassifier
from sklearn.pipeline import Pipeline
xgb_best = random_search_xgb.best_estimator_.named_steps['model']
xtrees_best = random_search_xtrees.best_estimator_.named_steps['model']
light_best = random_search_lgbm.best_estimator_.named_steps['model']

voting_clf = VotingClassifier(
    estimators=[
        ('xgb', xgb_best),
        ('xtrees', xtrees_best),
        ('lgbm', light_best)
    ],
    voting='soft',
    n_jobs=-1
)


final_pipe = ImbPipeline([
    ('preprocessor', preprocessor),
    ('oversampler',RandomOverSampler(random_state=42)),
    ('model', voting_clf)
])


final_pipe.fit(X_train, y_train)
y_pred_voting = final_pipe.predict(X_test)

print("Classification Report: Soft Voting Classifier (Simplified & Fixed)")
print("="*60)
print(classification_report(y_test, y_pred_voting))
````

## **6.5 Tuned Model Comparison**

All tuned candidates are compared on the testing data under the same fixed decision threshold (0.48), so that the comparison reflects the models and not different cut-off points. Candidate thresholds (0.37 to 0.50) were reviewed on the test set. F2 is slightly higher at lower thresholds, so 0.48 was selected as a practical compromise between Recall and workload (see Section 9), not because it maximizes F2. As a result, test metrics may be slightly optimistic.

````python
from sklearn.metrics import recall_score, fbeta_score, precision_score, roc_auc_score, accuracy_score

# 1. Definisikan list model terbaik yang telah kita tuning beserta threshold optimalnya untuk F2-Score
final_models = {
    'Tuned XGBoost': {
        'pipeline': best_xgb,
        'threshold': 0.48
    },
    'Tuned LightGBM': {
        'pipeline': best_lgbm,
        'threshold': 0.48
    },
    'Tuned Extra Trees': {
        'pipeline': best_xtrees,
        'threshold': 0.48
    },
    'Soft Voting': {
        'pipeline': final_pipe,
        'threshold': 0.48
    }
}

benchmark_tuned_results = []

# 2. Hitung metrik evaluasi pada test set menggunakan optimal threshold
for name, info in final_models.items():
    pipe = info['pipeline']
    thresh = info['threshold']

    # Dapatkan probabilitas kelas positif (1)
    y_proba = pipe.predict_proba(X_test)[:, 1]

    # Klasifikasikan berdasarkan threshold optimal
    y_pred_adjusted = (y_proba >= thresh).astype(int)

    # Kalkulasi metrik
    rec = recall_score(y_test, y_pred_adjusted)
    f2 = fbeta_score(y_test, y_pred_adjusted, beta=2)
    prec = precision_score(y_test, y_pred_adjusted)
    auc = roc_auc_score(y_test, y_proba)
    acc = accuracy_score(y_test, y_pred_adjusted)

    benchmark_tuned_results.append({
        'Model': name,
        'Optimal Threshold': thresh,
        'Recall': rec,
        'F2 Score': f2,
        'Precision': prec,
        'Accuracy': acc,
        'ROC AUC': auc
    })

# 3. Tampilkan hasil dalam bentuk DataFrame tabel
tuned_benchmark_df = pd.DataFrame(benchmark_tuned_results).sort_values(by='F2 Score', ascending=False).reset_index(drop=True)
display(tuned_benchmark_df.style.format({
    'Optimal Threshold': '{:.3f}',
    'Recall': '{:.4f}',
    'F2 Score': '{:.4f}',
    'Precision': '{:.4f}',
    'Accuracy': '{:.4f}',
    'ROC AUC': '{:.4f}'
}))
````

### **6.6 Overfitting vs Underfitting Diagnostic**

Overfitting occurs when a model learns the training data too closely and performs materially worse on unseen data. Underfitting occurs when the model is too simple to capture the relevant patterns, resulting in weak performance on both training and unseen data.

For this project, a simple diagnostic compares F2 on the training data and the untouched test data at the fixed 0.48 threshold. A large train–test gap is a warning signal for overfitting; low scores on both sides suggest underfitting. This diagnostic is used for model monitoring, not for selecting a threshold on the test set.


````python
# Overfitting / underfitting diagnostic at the fixed comparison threshold
DIAGNOSTIC_THRESHOLD = 0.48

diagnostic_models = {
    'XGBoost': best_xgb,
    'LightGBM': best_lgbm,
    'Extra Trees': best_xtrees,
    'Soft Voting': final_pipe
}

diagnostic_rows = []

for name, pipe in diagnostic_models.items():
    train_proba = pipe.predict_proba(X_train)[:, 1]
    test_proba = pipe.predict_proba(X_test)[:, 1]

    train_pred = (train_proba >= DIAGNOSTIC_THRESHOLD).astype(int)
    test_pred = (test_proba >= DIAGNOSTIC_THRESHOLD).astype(int)

    diagnostic_rows.append({
        'Model': name,
        'Train F2': fbeta_score(y_train, train_pred, beta=2),
        'Test F2': fbeta_score(y_test, test_pred, beta=2),
        'Train Recall': recall_score(y_train, train_pred),
        'Test Recall': recall_score(y_test, test_pred),
        'F2 Gap': fbeta_score(y_train, train_pred, beta=2) - fbeta_score(y_test, test_pred, beta=2)
    })

diagnostic_df = pd.DataFrame(diagnostic_rows).sort_values('F2 Gap', ascending=False)
display(diagnostic_df.round(4))
````

#**8. Revenue At Risk**

Revenue At Risk Prioritization

The model predicts cancellation probability for each reservation.

To quantify potential financial exposure, cancellation probability is combined with estimated booking revenue.

Revenue At Risk is calculated as:

Revenue At Risk = Cancellation Probability × Booking Revenue

This allows the hotel to prioritize intervention efforts toward reservations with the highest expected financial impact.

Bookings are grouped into risk levels by their Revenue At Risk: Low (bottom 50%), Medium (50th to 90th percentile), and High (top 10%). The 0.48 threshold decides which bookings are flagged as likely cancellations, while Revenue At Risk decides the order in which flagged bookings are handled.

````python
# Final Soft Voting probability for business prioritization
FINAL_THRESHOLD = 0.48
final_proba = final_pipe.predict_proba(X_test)[:, 1]

X_test_featured = create_features(X_test)

risk_df = X_test_featured.copy()
risk_df['cancel_probability'] = final_proba

# Revenue should not be negative in a business exposure calculation.
risk_df['revenue_base'] = risk_df['booking_revenue'].clip(lower=0)
risk_df['revenue_at_risk'] = risk_df['revenue_base'] * risk_df['cancel_probability']

risk_df['risk_level'] = pd.cut(
    risk_df['revenue_at_risk'],
    bins=[-np.inf, risk_df['revenue_at_risk'].quantile(0.50), risk_df['revenue_at_risk'].quantile(0.90), np.inf],
    labels=['Low', 'Medium', 'High']
)

risk_df = risk_df.sort_values('revenue_at_risk', ascending=False)

top_risk_df = risk_df[
    ['booking_revenue', 'cancel_probability', 'revenue_at_risk', 'risk_level']
].head(10)

display(top_risk_df.round(4))
````

````python
top_risk_plot = top_risk_df.copy()
top_risk_plot['Label'] = (
    'Rev=' + top_risk_plot['booking_revenue'].clip(lower=0).round(0).astype(int).astype(str)
    + ' | P=' + (top_risk_plot['cancel_probability'] * 100).round(0).astype(int).astype(str) + '%'
)

plt.figure(figsize=(12,6))
sns.barplot(data=top_risk_plot, y='Label', x='revenue_at_risk')
plt.title('Top Revenue At Risk Bookings')
plt.xlabel('Expected Revenue Exposure')
plt.ylabel('Booking Revenue | Cancellation Probability')
plt.tight_layout()
plt.show()
````

### Business Interpretation

A booking with a lower cancellation probability can still have greater financial priority when its booking revenue is substantially higher. This is why Revenue At Risk is used together with probability rather than probability alone.

# **9. Business Cost Approach**

**Business Cost Validation**

The deployment threshold recommendation is further evaluated through a cost-sensitive financial analysis based on estimated business impact.

Instead of using a relative cost ratio, prediction errors are translated into monetary values using operational follow-up costs and estimated revenue loss.

This approach enables threshold selection to be evaluated not only from a machine learning perspective, but also from a business value perspective.

**Cost Assumptions for Threshold Evaluation**

To ensure consistency with the Business Cost Evaluation framework, the following assumptions are applied:

```text
False Positive Cost = $10.00
False Negative Cost = $101.83

Total Business Cost = (False Positives × $10.00) + (False Negatives × $101.83)

````python
# ==========================================================
# BUSINESS COST VALIDATION (USD-BASED COST FRAMEWORK)
# ==========================================================

FP_COST = 10.00       # Follow-up cost per False Positive
FN_COST = 101.83      # Revenue loss per False Negative (ADR)

cost_results = []

for threshold in np.arange(0.10, 0.91, 0.01):

    pred = (
        y_proba_soft_voting >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        pred
    ).ravel()

    fp_cost = fp * FP_COST
    fn_cost = fn * FN_COST

    total_business_cost = (
        fp_cost +
        fn_cost
    )

    cost_results.append([
        threshold,
        fp,
        fn,
        fp_cost,
        fn_cost,
        total_business_cost
    ])

cost_df = pd.DataFrame(
    cost_results,
    columns=[
        'threshold',
        'false_positive',
        'false_negative',
        'fp_cost_usd',
        'fn_cost_usd',
        'total_business_cost_usd'
    ]
)

# ==========================================================
# BEST THRESHOLD BY LOWEST BUSINESS COST
# ==========================================================

best_cost_df = (
    cost_df
    .sort_values(
        by='total_business_cost_usd',
        ascending=True
    )
    .head(1)
)

display(
    best_cost_df.style.format({
        'threshold': '{:.2f}',
        'fp_cost_usd': '${:,.2f}',
        'fn_cost_usd': '${:,.2f}',
        'total_business_cost_usd': '${:,.2f}'
    })
)

print("=" * 70)
print("LOWEST BUSINESS COST THRESHOLD")
print("=" * 70)

best_row = best_cost_df.iloc[0]

threshold = best_row["threshold"]
fp = int(best_row["false_positive"])
fn = int(best_row["false_negative"])
fp_cost = best_row["fp_cost_usd"]
fn_cost = best_row["fn_cost_usd"]
total_cost = best_row["total_business_cost_usd"]

print("=" * 70)
print("LOWEST BUSINESS COST THRESHOLD")
print("=" * 70)

print(f"Threshold           : {threshold:.2f}")
print(f"False Positive      : {fp}")
print(f"False Negative      : {fn}")
print(f"FP Cost             : ${fp_cost:,.2f}")
print(f"FN Cost             : ${fn_cost:,.2f}")
print(f"Total Business Cost : ${total_cost:,.2f}")

print("=" * 70)
````

````python
# ==========================================================
# COST COMPARISON
# BUSINESS OPTIMUM vs DEPLOYMENT THRESHOLD
# ==========================================================

comparison_thresholds = [0.24, 0.48]

comparison_results = []

for th in comparison_thresholds:

    pred = (
        y_proba_soft_voting >= th
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        pred
    ).ravel()

    fp_cost = fp * FP_COST
    fn_cost = fn * FN_COST

    total_cost = fp_cost + fn_cost

    comparison_results.append({
        "Threshold": th,
        "FP": fp,
        "FN": fn,
        "FP Cost ($)": round(fp_cost, 2),
        "FN Cost ($)": round(fn_cost, 2),
        "Total Cost ($)": round(total_cost, 2)
    })

comparison_df = pd.DataFrame(comparison_results)

display(comparison_df)
````

## Business Cost Interpretation

To complement the machine learning evaluation metrics, a business cost analysis was conducted to quantify the financial impact of different deployment thresholds.

The analysis uses the following business assumptions:

```text
False Positive Cost = $10.00
False Negative Cost = $101.83
```

Under this framework, missing a cancellation is substantially more expensive than performing an unnecessary customer follow-up because a missed cancellation may result in lost revenue and missed mitigation opportunities.

### Financially Optimal Threshold

The lowest estimated business cost is achieved at **Threshold 0.24**, with the following results:

```text
False Positives  : 5,119
False Negatives  : 109

FP Cost          : $51,190.00
FN Cost          : $11,099.47

Total Cost       : $62,289.47
```

This threshold minimizes missed cancellations and therefore produces the lowest projected financial impact.

### Recommended Deployment Threshold

The selected deployment threshold (**0.48**) produces:

```text
False Positives  : 2,773
False Negatives  : 549

FP Cost          : $27,730.00

```

Although the total estimated business cost is higher than Threshold 0.24, the number of unnecessary interventions is reduced significantly, lowering operational workload by more than 2,300 bookings.

### Business Decision Rationale

From a financial perspective, **Threshold 0.24** represents the cost-optimal solution.

However, real-world deployment decisions should not be based solely on cost minimization. A highly aggressive threshold may generate an excessive number of customer follow-ups, increasing operational burden and reducing implementation practicality.

Therefore, **Threshold 0.48** is selected as the recommended production threshold because it provides a more balanced trade-off between:

- Revenue protection
- Cancellation detection capability
- Recall & F2-Score performance
- Operational efficiency
- Real-world business feasibility

In other words, the financially optimal threshold is not always the operationally optimal threshold. The final deployment decision prioritizes a sustainable balance between predictive performance, business impact, and operational capacity.

## Model and Threshold

The final model is a **Soft Voting ensemble** at a **0.48 threshold**.

- **Soft Voting ensemble (XGBoost + LightGBM + Extra Trees):** three prediction models whose probabilities are averaged into one final answer, similar to asking three experts and combining their opinions.
- **Threshold (0.48):** the minimum cancellation probability at which a booking is flagged as high risk. A booking with a probability of 48% or more is flagged.

The model was tested on 17,478 bookings it had never seen, of which 4,804 were actual cancellations:

| Metric | Value | What it means |
|---|---:|---|
| Recall | 0.8857 | Of all real cancellations, about 89 out of 100 were caught |
| Precision | 0.6054 | Of all flagged bookings, about 6 out of 10 really cancelled |
| F2 Score | 0.8107 | One overall score that combines Recall and Precision, giving more weight to Recall |
| Accuracy | 0.8099 | About 81 out of 100 bookings were classified correctly |
| ROC-AUC | 0.9166 | How well the model separates cancelling from non-cancelling bookings (1 is perfect, 0.5 is guessing) |

In counts, the model detected **4,255** cancellations, missed **549** (false negatives), and raised **2,773** false alarms (false positives: flagged bookings that did not cancel).

The train–test F2 gap is small (0.8349 vs. 0.8107), so there is no sign of serious **overfitting**, which is when a model memorizes its training data and performs poorly on new data. The top three tuned models are within 0.004 F2 of each other (XGBoost 0.8141, LightGBM 0.8110, Soft Voting 0.8107), so Soft Voting was chosen for stability rather than a clear accuracy advantage.

**Why threshold 0.48.** The business cost analysis assumes \$10 per false positive (a wasted follow-up) and \$101.83 per false negative (a missed cancellation). The lowest cost occurs at **threshold 0.24** (\$62,289) versus **0.48** (\$83,635). Threshold 0.24 would save about \$21,345 on paper, but it requires 2,346 more unnecessary follow-ups. Threshold 0.48 was chosen as the more practical balance between revenue protection and team workload.

## Future Work

- **Feature checks:** run an **ablation** (testing the model with and without a feature to see whether it truly helps) on the engineered features. `booking_revenue` is derived from `adr` (average daily rate) × nights, which are already predictors. Also replace `arrival_date_year` with month or week-of-year, since a year cannot generalize to the future.
- **Validation:** use a **time-based split** (train on older bookings, test on newer ones), select the threshold without the test set (e.g., from **out-of-fold predictions**, which are predictions on data the model did not train on), and **calibrate probabilities** so a predicted 70% really means about 70%. This makes Revenue At Risk more reliable.
- **Cost-aware modeling:** weight training by booking revenue, so the model pays more attention to valuable bookings, and set thresholds per booking from expected value instead of one global cut-off.
- **Advanced approaches:** **time-to-cancel (survival) modeling** predicts *when* a cancellation is likely, so outreach can be timed well. **Uplift modeling** finds guests who actually respond to an intervention, so effort is not spent on guests who would stay anyway.
- **Production:** benchmark a single tuned model against the ensemble (a simpler model may be easier to maintain), add **drift monitoring** (detecting when new bookings start to differ from the training data) and scheduled retraining, and bring in external data such as local events and competitor rates.

## Limitations

**Recall vs. precision trade-off.** The model is tuned toward Recall because a missed cancellation is assumed to cost more than an unnecessary follow-up. Catching more cancellations always means more false alarms.

| Threshold | Recall | Precision | Flagged | False alarms | Missed |
|---|---:|---:|---:|---:|---:|
| 0.24 | ≈ 0.977 | ≈ 0.478 | 9,814 | 5,119 | 109 |
| 0.48 | 0.886 | 0.605 | 7,028 | 2,773 | 549 |

A lower threshold protects more revenue but gives the reservation team many more bookings to contact. The 0.48 choice depends on assumed costs: the \$101.83 loss per missed cancellation equals the average daily rate, well below the average value of a cancelled booking (≈ \$478.06), so the framework is conservative. If real losses are higher or the team has more capacity, a lower threshold becomes easier to justify.

**Other limitations**
- The threshold cost analysis was computed on the test set, so test results may be slightly optimistic. The random split is also less realistic than a chronological one.
- Predicted probabilities are not calibrated, so Revenue At Risk is best read as a ranking signal, not an exact financial estimate.
- `country` and `agent` are the strongest features and may act as **proxies** (stand-ins) for specific markets or partners, which makes the model less stable if the booking mix changes.
- SHAP explains only the XGBoost component, not the full ensemble.
- The data is historical and covers two hotels. Revenue is estimated as ADR × nights, and cost figures are assumptions.
- The model predicts who may cancel, not whether contacting a guest will change the outcome, so each action should be validated in a pilot.
